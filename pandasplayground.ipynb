Sort by height ascending and weight descending

```python
birthyearandstatepeople.sort_values(
    by =['weight', 'height'],  # Columns to sort by
    ascending=[False, True]     # weight DESC, height ASC
)
```

```python
birthyearandstatepeople[
    birthyearandstatepeople['throws'] == 'R'
].sort_values(
    by=['weight', 'height'],  # Columns to sort by
    ascending=[False, False]     # weight DESC, height ASC
)
```

```python
Batting = pd.read_csv('data/Batting.csv')

HR_leaders = (
    Batting[['playerID', 'HR']]                    # select(playerID, HR)
    .groupby('playerID')                           # group_by(playerID)
    .agg(seasons_played=('HR', 'count'),           # summarise(seasons_played = n()
         total_HR=('HR', 'sum'),                   # sum(HR)
         mean_HR=('HR', 'mean'),                   # mean(HR)
         max_HR=('HR', 'max'))                     # max(HR)
    .reset_index()                                 # ungroup()
    .query('seasons_played > 20')                  # filter(seasons_played > 20)
)
```
